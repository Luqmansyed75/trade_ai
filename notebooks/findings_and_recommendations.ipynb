## 🔬 Part 2 — Extended Statistical Findings

The baseline Pearson test missed a crucial insight. By applying more rigorous non-parametric tests, we uncovered hidden relationships:

| Test | Statistic | p-value | Verdict |
|---|---|---|---|
| Pearson r | 0.0373 | 0.4152 | ✗ Not significant — looks for linear relationship only |
| **Spearman ρ** | **0.0997** | **0.0292** | **✓ Significant** — sentiment has a monotonic effect on PnL |
| **Kruskal-Wallis H** | **730.33** | **≈ 0.0000** | **✓ Highly significant** — PnL distributions fundamentally differ across sentiment groups |

> **Interpretation:** Market sentiment *does* affect trading outcomes. It changes the *shape and consistency* of profits, not just the average. A basic linear test was too simplistic to detect this.

## ⚖️ Part 4 — Risk-Adjusted Performance

Raw PnL doesn't tell the full story. By measuring how much *risk* was taken per dollar of *reward*, we find the safest trading environments:

| Sentiment | Reward/Risk | Profit Factor | Sharpe-like | Max Drawdown |
|---|---|---|---|---|
| Extreme Fear | 0.675 | 2.162 | 0.0436 | -$217,850 |
| Fear | 0.969 | 6.656 | 0.0838 | -$113,384 |
| Neutral | 0.924 | 4.321 | 0.0958 | -$132,639 |
| Greed | 0.911 | 3.031 | 0.0542 | -$460,774 |
| **Extreme Greed** | **1.339** | **11.023** | **0.1231** | -$152,323 |

> **Key Insight:** **Extreme Greed** is the most capital-efficient environment by every metric. A Profit Factor of 11.0 means traders earn $11 for every $1 they lose. The Reward/Risk ratio (1.34) confirms that average wins are significantly larger than average losses.

## 🔮 Part 6 — Lag-1 Predictive Analysis

Can yesterday's sentiment predict today's trading performance?

## 🚀 Strategy Recommendations

1. **Implement Directional Bias:** Do not trade longs and shorts equally. Heavily weight **Longs during Extreme Fear** and **Shorts during Extreme Greed** to trade against the crowd.

2. **Trade the Fear → Greed Breakout:** This transition yields the highest per-trade profit ($523). Actively scan for this specific day-to-day shift and scale into positions when it occurs.

3. **Optimize Capital by Risk:** Scale up leverage and position sizing during **Extreme Greed** (specifically on short setups) due to its mathematically superior Reward/Risk ratio (1.34) and Profit Factor (11.0).

4. **Avoid Lag-1 Predictions:** Yesterday's sentiment score alone does not significantly predict today's PnL. Use sentiment as a macro-level **environmental filter**, not an immediate buy/sell trigger.

5. **Study Top Traders:** The best wallet earned $2.14M total PnL. Analyse their behaviour across sentiment phases for deeper alpha.
